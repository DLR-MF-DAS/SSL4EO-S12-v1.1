## Fetching Data to Local Disk

```python
dataset = ssl4eos12_dataset.build_ssl4eos12_dataset(
    path="https://huggingface.co/datasets/embed2scale/SSL4EO-S12-v1.1/resolve/main/",
    modalities=["S2L2A", "S2L1C", "S2RGB", "S1GRD", "DEM", "NDVI", "LULC"], 
    split="val",
    batch_size=8,    
)
dataset
```

```python
import itertools
import tqdm.auto
n_samples = 3

ssl4eo_s12_v11_samples = [
    sample
    for sample in tqdm.auto.tqdm(itertools.islice(dataset, n_samples))
]
```

```python
# check data downloaded
len(ssl4eo_s12_v11_samples), \
ssl4eo_s12_v11_samples[0].keys(), \
ssl4eo_s12_v11_samples[0]['DEM']
```

## Data Visualization

```python
import numpy
import matplotlib
from matplotlib import pyplot as plt
```

```python
n_sample = 0
```

```python
# RGB data
vv, vh = \
    ssl4eo_s12_v11_samples[n_sample]['S1GRD'][0,0,0,...], \
    ssl4eo_s12_v11_samples[n_sample]['S1GRD'][0,0,1,...]
r = (vv - vv.min()) / (vv.max() - vv.min())
b = (vh - vh.min()) / (vh.max() - vh.min())
g = (r + b) / 2.

plt.imshow(ssl4eo_s12_v11_samples[n_sample]['S2RGB'][0,0,...].permute(1,2,0))
plt.title('Sentinel-2 RGB');
```

```python
# SAR data
vv, vh = \
    ssl4eo_s12_v11_samples[n_sample]['S1GRD'][0,0,0,...], \
    ssl4eo_s12_v11_samples[n_sample]['S1GRD'][0,0,1,...]
r = (vv - vv.min()) / (vv.max() - vv.min())
b = (vh - vh.min()) / (vh.max() - vh.min())
g = (r + b) / 2.

plt.imshow(numpy.stack([r,g,b], axis=2, dtype=float));
plt.title('Sentinel-1 SAR');
```

```python
# terrain data
plt.imshow(
    ssl4eo_s12_v11_samples[n_sample]['DEM'][0,0,0,...],
    cmap='terrain',
)
plt.title('terrain elevation');
```

```python
# land use data
plt.imshow(
    ssl4eo_s12_v11_samples[n_sample]['LULC'][0,0,0,...],
    cmap = matplotlib.colors.ListedColormap(numpy.random.rand(100, 3))
)
plt.title('land cover segmentation');
```
